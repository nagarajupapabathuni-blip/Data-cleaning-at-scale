```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt


print("Loading Dataset...")

df = pd.read_csv('/content/netflix_titles.csv')

print("\nDataset Loaded Successfully!")


print("\nFirst 5 Rows:")
print(df.head())

print("\nDataset Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns)

print("\nDataset Information:")
print(df.info())


print("\nMissing Values in Each Column:")
print(df.isnull().sum())


duplicate_rows = df.duplicated().sum()

print("\nDuplicate Rows:", duplicate_rows)

df.drop_duplicates(inplace=True)

print("Duplicates Removed Successfully!")


print("\nHandling Missing Values...")

df['director'] = df['director'].fillna('Unknown Director')

df['cast'] = df['cast'].fillna('Cast Not Available')

df['country'] = df['country'].fillna('Country Not Mentioned')

df['rating'] = df['rating'].fillna('Not Rated')

df['duration'] = df['duration'].fillna('Duration Unknown')

print("Missing Values Filled Successfully!")


print("\nConverting Date Format...")

df['date_added'] = pd.to_datetime(df['date_added'], format='mixed')

print("Date Format Converted!")


print("\nCleaning Column Names...")

df.columns = df.columns.str.strip()

print("Column Names Cleaned!")

print("\nFinal Missing Values:")
print(df.isnull().sum())


print("\nTop 10 Countries:")
print(df['country'].value_counts().head(10))

print("\nContent Type Count:")
print(df['type'].value_counts())


print("\nGenerating Visualization...")

df['type'].value_counts().plot(kind='bar')

plt.title('Movies vs TV Shows')

plt.xlabel('Content Type')

plt.ylabel('Count')

plt.show()

df.to_csv('cleaned_netflix_dataset.csv', index=False)

print("\nCleaned Dataset Saved Successfully!")


plt.figure(figsize=(10,5))

df['rating'].value_counts().head(10).plot(kind='bar')

plt.title('Top 10 Content Ratings')

plt.xlabel('Ratings')

plt.ylabel('Count')

plt.show()

plt.figure(figsize=(12,5))

df['country'].value_counts().head(10).plot(kind='bar')

plt.title('Top 10 Countries with Netflix Content')

plt.xlabel('Country')

plt.ylabel('Number of Shows')

plt.show()

plt.figure(figsize=(12,5))

df['release_year'].value_counts().head(15).sort_index().plot(kind='line')

plt.title('Content Release Trend')

plt.xlabel('Release Year')

plt.ylabel('Count')

plt.show()
```
